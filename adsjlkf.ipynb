```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

images = []

for i in range (9):
    img = Image.open (f"noisy_data/noisy_frame_00{i+1}.png")
    images.append (np.array(img))

stacked_cube = np.stack(images, axis = 0)

clean_array = np.median(stacked_cube, axis = 0)

clean_array = clean_array.astype (np.uint8)

plt.imshow (clean_array)
```
