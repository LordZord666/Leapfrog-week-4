```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open ('forinvert.jpg')

img_gray = img.convert('L') 

clean_img_array = np.array(img_gray)

img_int = clean_img_array.astype(int) #not necessary to convert
invert = np.invert(img_int)

plt.imshow(invert, cmap='gray')
plt.title("inverted img")
plt.axis ('off')
plt.show()
```
