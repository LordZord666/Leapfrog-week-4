```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open ('edge_detection.jpeg')

img_gray = img.convert('L') 

img_gray.save('grayed.jpg')
clean_img_array = np.array(img_gray)

img_int = clean_img_array.astype(int)
diff_y = np.abs(img_int[:-1,:] - img_int[1:,:])

diff_x = np.abs(img_int[:,:-1] - img_int[:,1:])

plt.imshow(diff_y, cmap='gray')
plt.title("vertical edges")
plt.axis ('off')
plt.show()
```

```python
plt.imshow(diff_x, cmap='gray')
plt.title("Horizontal edges")
plt.axis ('off')
plt.show()
```

```python
clean_img_array  = np.array(img)
flat_img = clean_img_array.flatten()

counts, bins = np.histogram(
    flat_img, bins = 256 , range = (0,256)
)

plt.plot (counts, color = "black")
plt.title ('Pixel intensity histogram')
plt.xlabel ("Pixel intensity (0-255)")
plt.ylabel ("Frequency")
plt.show()
```

```python
diff = np.maximum(diff_x[:-1,:], diff_y[:,:-1])

plt.imshow(diff, cmap='gray')
plt.title("edges")
plt.axis ('off')
plt.show()
```
