```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open('sample.jpg')

print(f"Format:(img.format)")
print(f"Size:(img.size)")
print(f"Mode:(img.mode)")

plt.imshow (img)
plt.axis("off")
plt.show()
```

```python
# images as numpy array

import numpy as np
img_array = np.array(img)

print(img_array.shape)

cropped = img_array[100:500,200:600,:]

plt.imshow (cropped)
plt.show()
```

```python
img_resized = img.resize((128,128))

img_gray = img_resized.convert('L') #pillow ko functionality of lightness use gareko 

img_gray.save('processed.jpg')

plt.imshow(img_gray, cmap = 'gray')
plt.axis('off')
plt.show() 
```

```python
import numpy as np

img_resized = img.resize ((128,128))

img_gray_array = np.mean (img_resized, axis = -1)

plt.imshow(img_gray_array, cmap = 'gray')
plt.axis('off')
plt.show()
```
